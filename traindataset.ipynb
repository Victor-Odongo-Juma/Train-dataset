```python
import pandas as pd
import numpy as np
```

```python
# Load the Titanic training dataset
df = pd.read_csv("C:/Users/Administrator/Desktop/Titanic datasets/train.csv")
```

```python
# Step 1.Display the first five observations
print(df.head())
```

```python
# Step 2.Examine Dataset Dimensions and Column Names
print("Dataset shape:", df.shape)

print("\nNumber of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())
```

```python
#Interpretation: The dataset contains 891 observations and 12 variables. Each row represents a passenger, while each column represents a recorded attribute.
```

```python
#Step 3: Inspect Data Types and General Information
print("Column data types:")
print(df.dtypes)

print("\nGeneral dataset information:")
df.info()
```

```python
#Step 4: Identify Missing Values
# Count missing values in each column
missing_count = df.isnull().sum()

# Calculate percentage of missing values
missing_percentage = (
    df.isnull().mean() * 100
)

# Create a summary table
missing_summary = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_percentage
})

print(
    missing_summary
    .sort_values("Missing Count", ascending=False)
    .round(2)
)
```

```python
#5: Check for Duplicate Records
# Count completely duplicated rows
print("Duplicate rows:", df.duplicated().sum())

# Check duplicate passenger identifiers
print(
    "Duplicate Passenger IDs:",
    df["PassengerId"].duplicated().sum()
)
```

```python
#Step 6: Generate Descriptive Statistics
# Numerical descriptive statistics
print(df.describe().round(2))

# Categorical descriptive statistics
print(df.describe(include=["object"]))
```

```python
#Step 7: Check for Invalid or Unexpected Values
# Check impossible numerical values
print("Negative ages:", (df["Age"] < 0).sum())
print("Negative fares:", (df["Fare"] < 0).sum())

# Inspect categorical values
print("\nSurvival outcomes:")
print(df["Survived"].value_counts(dropna=False))

print("\nPassenger classes:")
print(df["Pclass"].value_counts(dropna=False))

print("\nEmbarkation ports:")
print(df["Embarked"].value_counts(dropna=False))

# Check values outside expected categories
print(
    "\nUnexpected survival values:",
    (~df["Survived"].isin([0, 1])).sum()
)

print(
    "Unexpected passenger classes:",
    (~df["Pclass"].isin([1, 2, 3])).sum()
)
```

```python
#Step 8: Summarise the Data Quality Findings
print("DATA QUALITY REPORT")
print("-------------------")

print("Total observations:", len(df))
print("Total variables:", df.shape[1])

print(
    "Columns containing missing values:",
    df.isnull().any().sum()
)

print(
    "Completely duplicated rows:",
    df.duplicated().sum()
)

print(
    "Duplicate Passenger IDs:",
    df["PassengerId"].duplicated().sum()
)

print(
    "Columns with more than 30% missing:",
    df.columns[
        df.isnull().mean() > 0.30
    ].tolist()
)
```
